# Autoscaler Policy Parameters

This README explains the parameters used in the autoscaler policy for dynamically scaling AI inference instances based on resource usage and demand. The policy ensures optimal utilization and cost efficiency by automatically adjusting the number of running instances.

## Core Autoscaler Parameters

### `queue_up_threshold`
Queue length threshold to trigger scaling up (adding instances).
- **Default**: `10`
- **Example**: `20` - Scale up when average queue length exceeds 20 requests.

### `queue_down_threshold`
Queue length threshold to trigger scaling down (removing instances).
- **Default**: `2`
- **Example**: `1` - Scale down when average queue length drops below 1 request.

### `idle_time_downscale_threshold`
Time in seconds an instance can be idle before being considered for removal.
- **Default**: `600` (10 minutes)
- **Example**: `1800` - Remove instances idle for more than 30 minutes.

## Instance Limits

### `min_replicas`
Minimum number of instances to maintain.
- **Default**: `1`
- **Example**: `2` - Always keep at least 2 instances running.

### `max_replicas`
Maximum number of instances allowed.
- **Default**: `3`
- **Example**: `10` - Allow up to 10 instances during peak load.

## Timing and Averaging

### `averaging_period`
Time period for averaging queue length metrics.
- **Default**: `"average_1m"`
- **Example**: `"average_5m"` - Use 5-minute averages instead of 1-minute.

### `cooldown_seconds`
Time to wait after a scaling action before allowing another.
- **Default**: `120` (2 minutes)
- **Example**: `300` - Wait 5 minutes between scaling actions.

## Behavior Controls

### `allow_downscale_with_jobs`
Whether to allow scaling down instances that still have jobs in queue.
- **Default**: `true`
- **Example**: `false` - Never scale down instances with pending jobs.

## Example Configuration

```json
{
  "queue_up_threshold": 15,
  "queue_down_threshold": 3,
  "idle_time_downscale_threshold": 1200,
  "min_replicas": 1,
  "max_replicas": 5,
  "averaging_period": "average_1m",
  "cooldown_seconds": 180,
  "allow_downscale_with_jobs": false
}
```

This configuration scales up when queue exceeds 15, down when below 3, removes idle instances after 20 minutes, maintains 1-5 instances, uses 1-minute averages, waits 3 minutes between actions, and never scales down instances with jobs.
