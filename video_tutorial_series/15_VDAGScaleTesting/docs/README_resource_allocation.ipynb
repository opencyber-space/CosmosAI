# Resource Allocation Policy Parameters

This README explains the parameters used in the resource allocation policy for deploying instances across the cluster. The policy ensures each instance meets minimum hardware requirements and balances resource usage based on weighted metrics.

## Core Resource Parameters

### `min_cpu_per_instance`
Minimum number of vCPU required per instance.
- **Default/Example**: `5`
- **Purpose**: Ensures each instance has enough vCPU for AI workloads.

### `min_memory_per_instance`
Minimum RAM (in MB) required per instance.
- **Default/Example**: `3000`
- **Purpose**: Guarantees sufficient memory for model inference and data processing.

### `min_gpu_memory_per_instance`
Minimum GPU memory (in MB) required per instance.
- **Default/Example**: `47000`
- **Purpose**: Ensures each instance can load and run large AI models.

### `min_storage`
Minimum disk storage (in MB) required per instance.
- **Default/Example**: `47000`
- **Purpose**: Provides enough space for model files, logs, and temporary data.

### `gpu_memory_buffer`
Fraction of GPU memory to add as buffer.
- **Default/Example**: `0.1` (10%)
- **Purpose**: helps in preventing mistakes we have done during measuring actual GPU RAM requirement.

## Node Thresholds

### `node_cpu_threshold`
Maximum allowed CPU usage per node before considering this node for allocation.
- **Default/Example**: `0.8` (80%)
- **Purpose**: Avoids assigning tasks to nodes nearing CPU saturation.

### `node_memory_threshold`
Maximum allowed memory usage per node before  considering this node for allocation.
- **Default/Example**: `0.8` (80%)
- **Purpose**: Prevents memory exhaustion and swapping.

### `node_storage_threshold`
Maximum allowed storage usage per node before considering this node for allocation.
- **Default/Example**: `0.8` (80%)
- **Purpose**: Ensures disk space is available for new workloads.

## Resource Weights

### `weights`
Relative importance of each resource type in allocation decisions. Must sum to 1.0.
- **Default/Example**:
    ```json
    {
        "gpu_memory": 0.45,
        "cpu": 0.25,
        "memory": 0.2,
        "storage": 0.1
    }
    ```
- **Purpose**: Prioritizes GPU memory, then CPU, memory, and storage when selecting instances.

## Example Configuration

```json
{
  "min_cpu_per_instance": 5,
  "min_memory_per_instance": 3000,
  "min_gpu_memory_per_instance": 47000,
  "min_storage": 47000,
  "gpu_memory_buffer": 0.1,
  "node_cpu_threshold": 0.8,
  "node_memory_threshold": 0.8,
  "node_storage_threshold": 0.8,
  "weights": {
    "gpu_memory": 0.45,
    "cpu": 0.25,
    "memory": 0.2,
    "storage": 0.1
  }
}
```

This configuration ensures each instance meets strict hardware requirements, avoids overloading nodes, and prioritizes GPU memory in resource allocation decisions.
